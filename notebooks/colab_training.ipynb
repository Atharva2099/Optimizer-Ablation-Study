## 3. Import Project Modules
Here we verify that the model can be imported successfully from the `src` directory.

```python
import sys
import torch

# Add the src directory to the Python path
sys.path.append(os.path.abspath("src"))

from model import GPT

# Create a small "Tiny" configuration for debugging (as per plan.md)
class GPTConfig:
    vocab_size = 50257
    block_size = 256
    n_layer = 6
    n_head = 6
    n_embd = 384
    dropout = 0.0
    bias = False

config = GPTConfig()
model = GPT(config)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

params = sum(p.numel() for p in model.parameters()) / 1e6
print(f"Model instantiated with {params:.2f}M parameters on {device}.")
```

## 4. Training Loop Setup
*(Skeleton for training. You will import from `train.py` once it is implemented)*

```python
# Example placeholder for when your dataset and training loop are ready:
# from data import get_batch
# from train import train_step

optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=0.1)

# Dummy training step to verify backward pass works
print("Running a dummy forward/backward pass...")
dummy_x = torch.randint(0, config.vocab_size, (4, config.block_size)).to(device)
dummy_y = torch.randint(0, config.vocab_size, (4, config.block_size)).to(device)

logits, loss = model(dummy_x, dummy_y)
loss.backward()
optimizer.step()

print(f"Dummy step successful. Loss: {loss.item():.4f}")
```
